### Variáveis discretas e contínuas
Variáveis discretas são aquelas que assumem valores específicos e distintos, como o número de alunos em uma sala de aula. Já as variáveis contínuas podem assumir qualquer valor dentro de um intervalo, como a altura de uma pessoa.

### Intervalos de confiança
Um intervalo de confiança é uma faixa de valores, calculada a partir dos dados da amostra, que provavelmente contém o valor verdadeiro do parâmetro da população. Por exemplo, um intervalo de confiança de 95% significa que há 95% de chance de que o intervalo contenha o valor verdadeiro do parâmetro.

### Distribuições
#### Distribuição Hipergeométrica
A distribuição hipergeométrica descreve a probabilidade de \(k\) sucessos em \(n\) extrações, sem reposição, de uma população de tamanho \(N\) que contém exatamente \(K\) sucessos. A fórmula é:
$$
P(X = k) = \frac{\binom{K}{k} \binom{N-K}{n-k}}{\binom{N}{n}}
$$

#### Distribuição Normal
A distribuição normal, também conhecida como distribuição gaussiana, é uma distribuição contínua que é simétrica em torno da média. A função densidade de probabilidade é dada por:
$$
f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$
onde \(\mu\) é a média e \(\sigma\) é o desvio padrão.

#### Distribuição Binomial
A distribuição binomial descreve a probabilidade de obter exatamente \(k\) sucessos em \(n\) tentativas de um experimento de Bernoulli. A fórmula é:
$$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}
$$
onde \(p\) é a probabilidade de sucesso em uma única tentativa.

## Fisher exact test
O teste exato de Fisher é um teste estatístico usado para determinar se há uma associação significativa entre duas variáveis categóricas. É particularmente útil para tabelas de contingência pequenas.
