# Aula 02 - Coleta e Análise de dados

### Pipeline de aprendizado de máquina
- Um pipeline é uma sequência estruturada de etapas que processam dados e treinam modelos de forma organizada e eficiente
- Garante que todas as etapas sejam executadas corretamente e na ordem adequada
- Melhora a reprodutibilidade, escalabilidade e manutenção do modelo facilitando ajustes
